```python
from datetime import date, timedelta
import numpy as np

# Definir a data inicial
data_inicial = date(2025, 3, 27)                # Data de início (28/03/2025)

# Definir os feriados nacionais, estaduais e municipais aplicáveis
feriados = {
    date(2025, 4, 9),                           # Aniversário de Cubatão (Municípal)
    date(2025, 4, 18),                          # Sexta feira SANTA (Nacional)
    date(2025, 4, 21),                          # Tiradentes (Nacional)
    date(2025, 5, 1),                           # Dia do Trabalho (Nacional)
}

# Contar 24 dias úteis
dias_uteis = 0
data_atual = data_inicial

while dias_uteis < 24:
    data_atual += timedelta(days=1)
    # Verificar se não é sábado, domingo ou feriado
    if data_atual.weekday() < 5 and data_atual not in feriados:
        dias_uteis += 1

# Resultado final
data_atual
```
