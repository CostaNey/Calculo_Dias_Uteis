```python
from datetime import date, timedelta
import numpy as np

# Definir a data inicial
data_inicial = date(2025, 4, 15)                # Data de início (15/04/2025)

# Definir os feriados nacionais, estaduais e municipais aplicáveis
feriados = {
    date(2025, 4, 29),                # Aniversário de IPATINGA (Municípal)
    date(2025, 5, 1),                 # Dia do Trabalho (Nacional)
}

# Contar 24 dias úteis
dias_uteis = 0
data_atual = data_inicial

while dias_uteis < 60:
    data_atual += timedelta(days=1)  # Avançar um dia
    # Verificar se a data atual é um feriado ou final de semana
    # Verificar se não é sábado, domingo ou feriado
    if data_atual.weekday() < 5 and data_atual not in feriados:
        dias_uteis += 1

# Resultado final
data_atual
```
